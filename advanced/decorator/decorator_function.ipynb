```python
def meu_decorator(func):
    def wrapper():
        print('antes')
        func()
        print('depois')
        
    return wrapper

def ola():
    print('dentro do ola')

        
```

```python
#ponto importante, aqui passamos a função ola como parametro, que dentro do wrapper ela é executada. 
#uma função adia a execução de algo, posso criar uma função e guardar em uma variavel e so executar quando passar '()'
#o exemplo do meu_decorator é que ele retorna uma função, que nao esta sendo executada na definição, entao quando eu chamo o meu_decorator e executo ele: meu_decorator() o retorno dele é uma função, que deve tambem ser executada.
#ex: chamando o decorator e pedindo para executar o retorno dele: meu_decorator(ola)() ou variable = meu_decorator(ola) e depois variable()
meu_decorator(ola)
```

```python
#aqui o meu_decorator(ola) é executado e salvo em uma varible, quando executo o meu_decorator(ola) ele me retorna uma função, lembrando que uma função sempre retorna algo, que é uma função é uma maneira de adiar uma execução
variable = meu_decorator(ola) 
variable()
```

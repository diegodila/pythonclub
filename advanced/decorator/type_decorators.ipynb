```python
class Pessoa:
    def __init__(self,nome='Diego'):
        self.nome = nome

    @staticmethod
    def ace(*args,**kwargs):
        print(args,kwargs)
Pessoa.ace(1,2,4)
Pessoa.ace(dd=2,sdds=4)
```

```python
#uma construcao simples de um decorator, aqui a maior questao foi o retorno do wrapper dentro do meu_decorator, o return sempre retorna um valor da função, nao podemos fazer um return wrapper(), pq ele executaria antes do esperado o wrapper e retornaria None, pq o wrapper nao tem retorno
#um decorator nada mais é do que:---> ola = meu_decorator(ola)
#um detalhe importante é que eu preciso do wrapper para adiar a execução, sempre que eu preciso adiar uma execução de algo eu envolvo em um função
#o wrapper é um convenção, mas eu preciso de uma função dentro de outra para ter um decorator
def meu_decorator(func):
    def wrapper():
        print('antes')
        func()
        print('depois')
    return wrapper
    
@meu_decorator
def ola():
    print('Oi')
    
def meu_decorator2(func):
    def wrapper():
        print('antes 2')
        func()
        print('depois 2')
    return wrapper
    
ola2 = meu_decorator2(ola)



def ola3():
    print('Oi3')
    
```

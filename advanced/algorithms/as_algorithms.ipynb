```python
def numeros_pares(numeros):
    for numero in numeros:
        if numero % 2 == 0:
            yield numero


def multiplicar_por_dez(numeros):
    for numero in numeros:
        yield numero * 10


def primeiros(numeros, quantidade):
    contador = 0

    for numero in numeros:
        if contador >= quantidade:
            break

        yield numero
        contador += 1


numeros = range(1, 1_000_001)

pares = numeros_pares(numeros)
multiplicados = multiplicar_por_dez(pares)
resultado = primeiros(multiplicados, 10)

print(list(resultado))
```

```python
from itertools import islice


def numeros_pares(numeros):
    """Produz somente números pares."""
    for numero in numeros:
        if numero % 2 == 0:
            yield numero


def multiplicar_por_dez(numeros):
    """Produz cada número multiplicado por dez."""
    for numero in numeros:
        yield numero * 10


def processar(numeros, quantidade):
    """Produz uma quantidade limitada de resultados."""
    yield from islice(
        multiplicar_por_dez(numeros_pares(numeros)),
        quantidade,
    )


numeros = range(1, 1_000_001)

resultado = list(processar(numeros, 10))

print(resultado)
```

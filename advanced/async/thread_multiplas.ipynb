```python
import threading
import time
```

```python
def main():
    threads = [threading.Thread(target=contar, args=('elefante', 10)),
               threading.Thread(target=contar, args=('sapo', 2)),
               threading.Thread(target=contar, args=('lion', 20)),
               threading.Thread(target=contar, args=('orange', 6)),
               ]
    
    #Adiciona a nossa thread na pool de therads prontas para execução
    [th.start() for th in threads]
    
    print('Podemos fazer outras coisas no programa enquanto a thread vai executando...')
    print('pythonclub\n' * 2)
    
    #avisa para ficar aguardando aqui (ate esse ponto do programa) até a thread terminar a execução
    [th.join() for th in threads]
    
    print('Done')
    
def contar(o_que, numero):
    for n in range(1,numero + 1):
        print(f'{n} {o_que}(s)...')
        time.sleep(2)
```

```python
if __name__ == '__main__':
    main()
```

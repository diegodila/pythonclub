```python
import threading
import time
```

```python
def main():
    th = threading.Thread(target=contar, args=('elefante ', 10))
    
    #Adiciona a nossa thread na pool de therads prontas para execução
    th.start() 
    
    print('Podemos fazer outras coisas no programa enquanto a thread vai executando...')
    print('pythonclub\n' * 2)
    
    #avisa para ficar aguardando aqui (ate esse ponto do programa) até a thread terminar a execução
    th.join()
    
    print('Done')
    
def contar(o_que, numero):
    for n in range(1,numero + 1):
        print(f'{n} {o_que}(s)...')
        time.sleep(2)
```

```python
if __name__ == '__main__':
    main()
```

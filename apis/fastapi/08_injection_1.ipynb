```python
from fastapi import FastAPI
```

```python
app = FastAPI()
```

```python
times = { 1: {"clube": "Palmeiras",
             "titulos": 25,
             },
         2: {"clube":"Corinthians", 
             "titulos": 19}
         }
```

```python
@app.get('/times')
def get_times():
    return times
```

```python
@app.get('/times/{numero}')
def get_time(numero:int):
    try:
        return times[numero]
    except:
        print('erro')
```

```python
@app.post('/times')
async def post_times(time):
    
    times[time.id] = time
    return time
```

```python
import uvicorn
from uvicorn import Config, Server

config = Config(app=app, host="0.0.0.0", port=8000)
server = Server(config)

await server.serve()
```

```python
import gspread
import pandas as pd
from oauth2client.service_account import ServiceAccountCredentials
from datetime import date, datetime, timedelta
import re
```

```python
# define the scope
scope = ['https://spreadsheets.google.com/feeds','https://www.googleapis.com/auth/drive']

# add credentials to the account, needed make download in google cloud platform 
creds = ServiceAccountCredentials.from_json_keyfile_name('./work/pythonclub/apis/cred.json', scope)

# authorize the clientsheet
client = gspread.authorize(creds)
```

```python
client.open('minhas_contas').sheet1.get()
```

```python
from google.oauth2.service_account import Credentials

scopes = [
    'https://www.googleapis.com/auth/spreadsheets',
    'https://www.googleapis.com/auth/drive'
]

credentials = Credentials.from_service_account_file(
    './work/pythonclub/cred.json',
    scopes=scopes
)

gc = gspread.authorize(credentials)
```

```python
gc.open('minhas_contas').sheet1.get()
```

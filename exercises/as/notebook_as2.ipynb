```python
class Carro:
    def __init__(self,marca:str=None):
        self.marca = marca

    @staticmethod
    def metodo_statico(*args, **kwargs):
        print(args, kwargs)
```

```python
c = Carro()
```

```python
c.metodo_statico(1,2,3,4)
```

```python
class Pessoa:
    def __init__(self, nome:str=None):
        self._nome = nome

    @property
    def get_nome(self):
        return self._nome
```

```python
class Pessoa:
    ano_atual = 2022
    def __init__(self,nome='Diego', idade=11):
        self.nome = nome
        self.idade = idade

    @classmethod
    def cria_instancia(cls,nome='Diego'):
        return cls(nome)
    
p = Pessoa('Df')
p1 = Pessoa().cria_instancia('dsdi')

p1.nome
```

```python
class Pessoa:
    def __init__(self, nome='Diego'):
        self.nome = nome

    @staticmethod
    def ace(*args,**kwargs):
        print(args,kwargs)
        
Pessoa.ace(1,2,4)
Pessoa.ace(dd=2,sdds=4)
```
